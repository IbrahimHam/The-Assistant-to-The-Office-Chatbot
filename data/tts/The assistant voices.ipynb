```python
import os
from glob import glob
from TTS.api import TTS
import torch
from TTS.tts.configs.xtts_config import XttsConfig
from TTS.tts.models.xtts import XttsAudioConfig
from TTS.config.shared_configs import BaseDatasetConfig

# Allowlist the required classes for safe unpickling
torch.serialization.add_safe_globals([XttsConfig, XttsAudioConfig, BaseDatasetConfig])

# Initialize the TTS model
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2", gpu=False)

# Path to the folder containing .wav files
wav_folder = "/Users/nikitabuss/Desktop/Spiced_Academy/The-Assistant-to-The-Office-Chatbot/tts/"

# Get all .wav files in the folder
wav_files = glob(os.path.join(wav_folder, "*.wav"))

# Check if there are any .wav files
if not wav_files:
    print("No .wav files found in the folder.")
else:
    for wav_file in wav_files:
        print(f"Processing: {wav_file}")
        # Generate speech for each .wav file
        tts.tts_to_file(
            text="David, here it is. My philosophy is basically this. And this is something that I live by. And I always have. And I always will. Don't ever, for any reason, do anything to anyone, for any reason, ever, no matter what. No matter... where. Or who, or who you are with, or, or where you are going, or... or where you've been... ever. For any reason, whatsoever.",
            file_path=f"output_{os.path.basename(wav_file)}",  # Save output with a unique name
            speaker_wav=wav_file,  # Use the current .wav file
            language="en"
        )
```
